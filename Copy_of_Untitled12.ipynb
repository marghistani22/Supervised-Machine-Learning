```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Problem 1: Cat Learning Curve
# Load data
data = pd.read_csv("/content/cats (1).txt", delimiter="\t")

# Scatterplot
plt.scatter(data['Trial'], data['rate'])
plt.xlabel('Number of Trials')
plt.ylabel('Escape Rate (escapes per minute)')
plt.title('Cat Learning Curve')
plt.show()

# Correlation
correlation = data['Trial'].corr(data['rate'])
print("Correlation:", correlation)

# Least-Squares Line
slope, intercept, _, _, _ = linregress(data['Trial'], data['rate'])
plt.scatter(data['Trial'], data['rate'], label='Data Points')
plt.plot(data['Trial'], slope * data['Trial'] + intercept, color='red', label='Least-Squares Line')
plt.xlabel('Number of Trials')
plt.ylabel('Escape Rate')
plt.legend()
plt.show()

# Residuals Plot
residuals = data['rate'] - (slope * data['Trial'] + intercept)
plt.scatter(data['rate'], residuals)
plt.xlabel('Escape Rate')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```
